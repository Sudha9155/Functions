# Regression Assignment

## Q1. What is Simple Linear Regression?
Answer:
Simple Linear Regression is a statistical technique used to model the relationship between one independent variable (X) and one dependent variable (Y). It predicts the value of Y using the equation:
Y = mX + c

---

## Q2. What are the key assumptions of Simple Linear Regression?
Answer:
The key assumptions are:
1. Linear relationship between X and Y.
2. Independence of observations.
3. Homoscedasticity (constant variance of errors).
4. Normally distributed residuals.
5. No significant outliers.

---

## Q3. What does the coefficient m represent in the equation Y = mX + c?
Answer:
The coefficient m is called the slope of the line. It represents the amount by which the dependent variable (Y) changes for a one-unit increase in the independent variable (X).

---

## Q4. What does the intercept c represent in the equation Y = mX + c?
Answer:
The intercept c is the value of Y when X is equal to zero. It represents the point where the regression line crosses the Y-axis.

---

## Q5. How do we calculate the slope m in Simple Linear Regression?
Answer:
The slope is calculated using the formula:

m = Σ[(Xi − X̄)(Yi − Ȳ)] / Σ[(Xi − X̄)²]

where X̄ and Ȳ are the mean values of X and Y respectively.

---

## Q6. What is the purpose of the least squares method in Simple Linear Regression?
Answer:
The least squares method finds the best-fit regression line by minimizing the sum of the squared differences between the actual values and the predicted values.

---

## Q7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?
Answer:
R² measures how well the regression model explains the variation in the dependent variable. Its value ranges from 0 to 1. A higher R² indicates a better model fit.

---

## Q8. What is Multiple Linear Regression?
Answer:
Multiple Linear Regression is an extension of Simple Linear Regression that uses two or more independent variables to predict a dependent variable.

Equation:
Y = b0 + b1X1 + b2X2 + ... + bnXn

---

## Q9. What is the main difference between Simple and Multiple Linear Regression?
Answer:
Simple Linear Regression uses one independent variable, whereas Multiple Linear Regression uses two or more independent variables to predict the dependent variable.

---

## Q10. What are the key assumptions of Multiple Linear Regression?
Answer:
The assumptions are:
1. Linear relationship between predictors and target.
2. Independence of observations.
3. Homoscedasticity.
4. Normal distribution of residuals.
5. No multicollinearity among independent variables.
6. No significant outliers.

## Q11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?
Answer:
Heteroscedasticity occurs when the variance of the residuals (errors) is not constant across all levels of the independent variables. It can make standard errors unreliable, leading to incorrect hypothesis tests and confidence intervals, even though the coefficient estimates remain unbiased.

---

## Q12. How can you improve a Multiple Linear Regression model with high multicollinearity?
Answer:
High multicollinearity can be reduced by:
1. Removing highly correlated features.
2. Combining correlated variables.
3. Using Principal Component Analysis (PCA).
4. Applying Ridge or Lasso Regression.
5. Collecting more data if possible.

---

## Q13. What are some common techniques for transforming categorical variables for use in regression models?
Answer:
Common techniques include:
1. Label Encoding.
2. One-Hot Encoding.
3. Ordinal Encoding.
4. Target Encoding.
5. Binary Encoding.
The choice depends on the type of categorical variable and the regression model.

---

## Q14. What is the role of interaction terms in Multiple Linear Regression?
Answer:
Interaction terms capture the combined effect of two or more independent variables on the dependent variable. They help model situations where the effect of one variable depends on the value of another variable.

---

## Q15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?
Answer:
In Simple Linear Regression, the intercept is the predicted value of the dependent variable when the independent variable is zero. In Multiple Linear Regression, it is the predicted value when all independent variables are zero simultaneously.

---

## Q16. What is the significance of the slope in regression analysis, and how does it affect predictions?
Answer:
The slope represents the expected change in the dependent variable for a one-unit increase in the independent variable while keeping other variables constant. It determines the direction and strength of the relationship and directly affects predictions.

---

## Q17. How does the intercept in a regression model provide context for the relationship between variables?
Answer:
The intercept provides the baseline value of the dependent variable when all independent variables are zero. It serves as the starting point of the regression equation and helps interpret the model.

---

## Q18. What are the limitations of using R² as a sole measure of model performance?
Answer:
Limitations of R² include:
1. It does not indicate whether predictions are accurate.
2. It always increases when more variables are added.
3. It cannot detect overfitting.
4. It does not evaluate model assumptions.
Therefore, R² should be used along with metrics such as Adjusted R², MAE, RMSE, and cross-validation.

---

## Q19. How would you interpret a large standard error for a regression coefficient?
Answer:
A large standard error indicates that the estimated coefficient is less precise and may vary significantly across different samples. It suggests greater uncertainty and may imply that the coefficient is not statistically significant.

---

## Q20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?
Answer:
Heteroscedasticity can be identified by plotting residuals against predicted values. If the spread of residuals changes (such as a funnel-shaped pattern), heteroscedasticity is present. Addressing it is important because it improves the reliability of statistical tests and confidence intervals, leading to more accurate model interpretation.

## Q21. What does it mean if a Multiple Linear Regression model has a high R² but low Adjusted R²?
Answer:
A high R² but low Adjusted R² indicates that some independent variables do not contribute meaningfully to the model. R² always increases when new variables are added, while Adjusted R² penalizes unnecessary variables. This may indicate overfitting.

---

## Q22. Why is it important to scale variables in Multiple Linear Regression?
Answer:
Feature scaling ensures that variables with different units or ranges are on a similar scale. It improves numerical stability, speeds up optimization, and is especially useful when using regularization methods such as Ridge and Lasso Regression.

---

## Q23. What is polynomial regression?
Answer:
Polynomial Regression is an extension of Linear Regression that models the relationship between the independent and dependent variables using polynomial terms. It is useful when the relationship between variables is non-linear.

---

## Q24. How does polynomial regression differ from linear regression?
Answer:
Linear Regression fits a straight line to the data, while Polynomial Regression fits a curved line by including higher-degree terms such as X², X³, etc. Polynomial Regression is suitable for non-linear relationships.

---

## Q25. When is polynomial regression used?
Answer:
Polynomial Regression is used when the relationship between variables is curved rather than linear. It is commonly applied in growth prediction, sales forecasting, engineering, and scientific data analysis.

---

## Q26. What is the general equation for polynomial regression?
Answer:
The general equation is:

Y = b0 + b1X + b2X² + b3X³ + ... + bnXⁿ

where b0, b1, b2, ..., bn are coefficients and n is the degree of the polynomial.

---

## Q27. Can polynomial regression be applied to multiple variables?
Answer:
Yes. Polynomial Regression can be extended to multiple independent variables by including polynomial and interaction terms for each variable.

---

## Q28. What are the limitations of polynomial regression?
Answer:
The limitations are:
1. It may overfit the training data.
2. It becomes complex for high-degree polynomials.
3. It is sensitive to outliers.
4. Choosing the correct polynomial degree is difficult.
5. It may not generalize well to unseen data.

---

## Q29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?
Answer:
Common methods include:
1. R² Score
2. Adjusted R²
3. Mean Squared Error (MSE)
4. Root Mean Squared Error (RMSE)
5. Cross-validation
These methods help select the best polynomial degree while avoiding overfitting.

---

## Q30. Why is visualization important in polynomial regression?
Answer:
Visualization helps understand the relationship between variables, identify whether a polynomial model is appropriate, detect overfitting or underfitting, and compare the fitted curve with the actual data.

---

## Q31. How is polynomial regression implemented in Python?
Answer:
Polynomial Regression in Python is implemented using the PolynomialFeatures class from sklearn.preprocessing to create polynomial features and LinearRegression from sklearn.linear_model to train the model. The model is then fitted using the transformed features and used for prediction.

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Sample Data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 5, 10, 17, 26])

# Create Polynomial Features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Train Model
model = LinearRegression()
model.fit(X_poly, y)

# Prediction
pred = model.predict(X_poly)

print("Predicted Values:")
print(pred)
```
